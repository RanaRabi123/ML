```python
import pandas as pd
import numpy as np
import os 
```

```python
class students():
    def __init__(self):
        try:
            self.file_path = r'D:\intern preparation\python_practice\Students_record.xlsx'
            self.st_df = pd.read_excel(self.file_path)
        except:
            st_df = pd.DataFrame(columns=['id', 'name', 'age', 'gender', 'class'])
            st_df.to_excel(self.file_path, index=False)
            if os.path.exists(self.file_path):
                self.st_df = pd.read_excel(self.file_path)
            else:
                raise FileNotFoundError('Error occured while loading file ... ')


    def add_student(self):
        self.st_df.loc[len(self.st_df)] = [int(len(self.st_df)+1) , input("Enter student name : "), int(input("Enter student age : ")), input('Enter student gender : '), input("Enter student class : ")]
            
        self.st_df.to_excel(self.file_path, index=False)
        print('\nStudent added successfully... \n', '--'*20)


    def update_student(self):
        user_id = int(input("Enter student id to update : "))
        while True:
            user_info_col = int(input("what you want to change : \n 1. Name \n 2. Age \n 3. Gender \n 4. Class \n 5. ...Exit..."))
            if user_info_col==5:
                break
            elif user_info_col>0 and user_info_col<5 and user_info_col!=2:
                self.st_df.at[user_id-1, self.st_df.columns[user_info_col]] = input('Enter new credential : ')
                self.st_df.to_excel(self.file_path, index=False)
            elif user_info_col==2:
                self.st_df.at[user_id-1, self.st_df.columns[user_info_col]] = int(input('Enter new credential : '))
                self.st_df.to_excel(self.file_path, index=False)
            else:
                print('Invalid choice, select within range...')
                continue
            
        print('\nStudent Updated successfully ... \n', '--'*20)


    def delete_student(self):
        user_id = int(input('Enter sudent id to drop : '))
        self.st_df.drop(index=user_id-1, inplace=True, axis=0)
        self.st_df.to_excel(self.file_path, index=False)

        print('\nStudent removed successfully ... \n', '--'*20)


    def search_student(self):
        user_id = int(input('Enter student id : '))
        print(self.st_df.loc[user_id-1], '\n','=='*20)
        
    def display_record(self):
        print(self.st_df.info())
        print('\n','=='*20, '\n',self.st_df, '\n', '=='*20)
```

```python
obj1 = students()
while True:
    user_choice = input("What you want to do : \n 1. Add Student \n 2. Update Student \n 3. Delete Student \n 4. Search Student \n 5. Display all records \n 6. Exit")
    if user_choice=='1':
        obj1.add_student()
    elif user_choice=='2':
        obj1.update_student()
    elif user_choice=='3':
        obj1.delete_student()
    elif user_choice=='4':
        obj1.search_student()
    elif user_choice=='5':
        obj1.display_record()
    elif user_choice=='6':
        print('--'*20,'\nStudent Management System closing ...')
        break
    else:
        print('Invalid choice. Select from within range.')
```
